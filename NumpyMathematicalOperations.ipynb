# Basic Arithmatic operations

```python
import numpy as np

a = np.array([19, 5, 88, 99, 93, 13])
b = np.array([2, 3, 4, 5, 6, 7])

addition = np.add(a, b)
subtraction = np.subtract(a, b)
multiplication = np.multiply(a, b)
division = np.divide(a, b)
modulus = np.mod(a, b)
exponential = np.power(a, b)

print("Addition: \n", addition)
print("Subtraction: \n", subtraction)
print("Multiplication: \n", multiplication)
print("Division: \n", division)
print("Modulus: \n", modulus)
print("Exponential: \n", exponential)
```

# numpy.sum() in Python

```python
import numpy as np

arr = np.array([5, 10, 15])
print(np.sum(arr))


# Example 1:

arr = np.array([10, 20, 2.0, 4, 2])

print(np.sum(arr))
print(np.sum(arr, dtype=np.uint8))
print(np.sum(arr, dtype=np.float32))


# Example 2

arr = np.array([[2, 45, 56, 7],
                [45, 68, 83, 88],
                [8, 2, 1, 3]])
print(np.sum(arr))
print(np.sum(arr, dtype=np.uint8))
print(np.sum(arr, dtype=np.float32))


# Example 3

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
print(np.sum(arr, axis=1, keepdims=True))
```

# numpy.mean() in Python

```python
import numpy as np

arr = np.array([2, 3, 7, 8, 5])
print(np.mean(arr))

# Example 1

arr = np.array([2, 8, 9, 6])
res = np.mean(arr)
print(res)


# Example 2

arr = np.array([[12, 45, 67.89],
                [23, 89, 78.3],
                [34, 45.3, 77]])

print(np.mean(arr))
print(np.mean(arr, axis=0))
print(np.mean(arr, axis=1))


# Example 3

res = np.zeros(3)
np.mean(arr, axis=1, out=res)
print(res)
```
