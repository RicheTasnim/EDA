# Basic Arithmatic operations

```python
import numpy as np

a = np.array([19, 5, 88, 99, 93, 13])
b = np.array([2, 3, 4, 5, 6, 7])

addition = np.add(a, b)
subtraction = np.subtract(a, b)
multiplication = np.multiply(a, b)
division = np.divide(a, b)
modulus = np.mod(a, b)
exponential = np.power(a, b)

print("Addition: \n", addition)
print("Subtraction: \n", subtraction)
print("Multiplication: \n", multiplication)
print("Division: \n", division)
print("Modulus: \n", modulus)
print("Exponential: \n", exponential)
```

# numpy.sum() in Python

```python
import numpy as np

arr = np.array([5, 10, 15])
print(np.sum(arr))


# Example 1:

arr = np.array([10, 20, 2.0, 4, 2])

print(np.sum(arr))
print(np.sum(arr, dtype=np.uint8))
print(np.sum(arr, dtype=np.float32))


# Example 2

arr = np.array([[2, 45, 56, 7],
                [45, 68, 83, 88],
                [8, 2, 1, 3]])
print(np.sum(arr))
print(np.sum(arr, dtype=np.uint8))
print(np.sum(arr, dtype=np.float32))


# Example 3

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
print(np.sum(arr, axis=1, keepdims=True))
```

# numpy.mean() in Python

```python
import numpy as np

arr = np.array([2, 3, 7, 8, 5])
print(np.mean(arr))

# Example 1

arr = np.array([2, 8, 9, 6])
res = np.mean(arr)
print(res)


# Example 2

arr = np.array([[12, 45, 67.89],
                [23, 89, 78.3],
                [34, 45.3, 77]])

print(np.mean(arr))
print(np.mean(arr, axis=0))
print(np.mean(arr, axis=1))


# Example 3

res = np.zeros(3)
np.mean(arr, axis=1, out=res)
print(res)
```

# numpy.maximum() in Python

```python
import numpy as np

a = 10
b = 20
print(np.maximum(a,b))


# Example 1

a = np.array([2, 18, 125])
b = np.array([3, 3, 15])
print(np.maximum(a, b))


# Example 2

a = np.array([np.nan, 0, np.nan])
b = np.array([np.nan, np.nan, 0])
print(np.maximum(a, b))

# Example 3

a = np.array([[1, 4, 7], [2, 5, 8]])
b = np.array([3, 3, 3])
print(np.maximum(a, b))
```

# numpy.minimum() in Python

```python
import numpy as np

a = 10
b = 21
print(np.minimum(a, b))

# Example 1

a = np.array([2, 8, 125])
b = np.array([3, 3, 15])
print(np.minimum(a, b))

# Example 2

a = np.array([np.nan, 0, np.nan])
b = np.array([np.nan, np.nan, 0])
print(np.minimum(a, b))

# Example 3

a = np.array([[4, 7, 9], [1, 5, 8]])
b = np.array([3, 6, 10])
print(np.minimum(a, b))
```

# Numpy unfuncs | Universal functions

```python
import numpy as np

# 1. Trigonometric functions

angles = np.array([0, 30, 45, 60, 90])
rad = np.deg2rad(angles) # converts degree to radius

# Sine of angles
sin_vals = np.sin(rad)
print("Sine Values: ", sin_vals)

# Inverse sine in degrees
inv_sin = np.rad2deg(np.arcsin(sin_vals))
print("Inverse sine (degrees): ", inv_sin)

# Hyperbolic sine
sinh_vals = np.sinh(rad)
print("Hyperbolic sine: ", sinh_vals)

# Hypotenuse of a right triangle
hyp = np.hypot(3, 4)
print("Hypotenuse: ", hyp)


# 2. Statistical functions

weights = np.array([50., 52.5, 50, 58, 55.63, 73.25, 49.5, 45])

# Min and Max
print("Min and Max: ", np.amin(weights), np.amax(weights))

# Range
print("Range: ", np.ptp(weights))

# 70 th percentile
print("70th percentile: ", np.percentile(weights, 70))

# Mean
print("Mean: ", np.mean(weights))

# Median
print("Median: ", np.median(weights))

# Standard Deviation
print("Standard Deviation: ", np.std(weights))

# Variance
print("Variance: ", np.std(weights))

# Average
print("Average: ", np.average(weights))


# 3. Bitwise Functions

even = np.array([0, 2, 4, 6, 8, 16, 32])
odd = np.array([1, 3, 5, 7, 9, 17, 33])

# Bitwise AND, OR, XOR
print("AND: ", np.bitwise_and(even, odd))
print("OR: ", np.bitwise_or(even, odd))
print("XOR: ", np.bitwise_xor(even, odd))

# Bitwise NOT
print("NOT: ", np.invert(even))

# Bit shifts
print("Left Shift: ", np.left_shift(even, 1))
print("Right Shift: ", np.right_shift(even, 1))
```
