# Matrix Multiplication

```python
import numpy as np

# Example 1

a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]
print("Matrix A: ", a, "\nMatrix B: ", b)

result = np.dot(a, b)
print("Result: \n", result)

# Example 2

c = [[1, 2], [3, 4], [5, 6]]
d = [[4, 3, 2], [7, 5, 6]]
print("Matrix C: ", c, "\nMatrix D: ", d)

result = np.dot(c, d)
print("Result: \n", result)
```

# Matrix Manipulation

```python
import numpy as np

# 1. Element-wise Addition, Subtraction, Division

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Addition: \n", np.add(a, b))
print("Subtraction: \n", np.subtract(a, b))
print("Division: \n", np.divide(a, b))
print("Element-wise operation: \n", np.multiply(a, b))
print("Matrix Multiplication: \n", np.dot(a, b))
print("Squre Root: \n", np.sqrt(a))
print("Sum of all elements: \n", np.sum(b))
print("Column-wise sum: ", np.sum(b, axis=0))
print("Row-wise sum: ", np.sum(a, axis=1))
print("Transpose: \n", a.T)
```

# Matrix and vector products

```python
import numpy as np

# Ineer Product

# Define vectors
a = np.array([2, 6])
b = np.array([4, 8])
print("Inner product of vectors: ", np.inner(a, b))

# Define Matrices
x = np.array([[1, 3, 7], [5, 9, 6]])
y = np.array([[22, 33, 46], [84, 45, 91]])
print("Inner product of Matrices: \n", np.inner(x, y))

# outer product
print("Outer product of vectors: ", np.outer(a,b))
print("Outer product of vectors: \n", np.outer(x, y))

# Cross Product
a = np.array([3, 5, 0])
b = np.array([7, 3, 0])

print("Cross product of vector: ", np.cross(a, b))
print("Cross product of matrices: \n", np.cross(x, y))
```

# Determinants

```python
import numpy as np

# numpy.linalg.slogder()
A = np.array([[50, 29], [30, 44]])
sign, logdet = np.linalg.slogdet(A)
res = sign * np.exp(logdet)
print(res)

# numpy.linalg.det()
A = np.array([[1, 2], [3, 4]])
res = np.linalg.det(A)
print(res)

# scipy.linalg.lu
import scipy.linalg
A = np.array([[1, 2], [3, 4]])
P, L, U = scipy.linalg.lu(A)
res = np.prod(np.diag(U))
print(res)
```

# Inverse a Matrix

```python
import numpy as np

# Example 1
A = np.array([[6, 1, 1],
              [4, -2, 5],
              [2, 8, 7]])
print(np.linalg.inv(A))

# Example 2
A = np.array([[6, 1, 1, 3],
              [4, -2, 5, 1],
              [5, 7, 4, 6],
              [8, 9, 2, 1]])
print(np.linalg.inv(A))

# Example 3
B = np.array([[[1, 2], [3, 4]],
              [[1, 3], [3, 5]]])
print(np.linalg.inv(B))
```
