# Types of Seaborn plots

Relational plots

```python
#Scatter plot
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
# Line plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.lineplot(data=tips, x="size", y="tip")
plt.show()
```

```python
# Relational plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.relplot(data=tips, x="total_bill", y="tip", hue="smoker")
plt.show()
```

Categorical Plot

```python
# Bar plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# Count plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=tips, x="day")
plt.show()
```

```python
# Box plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxenplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# Violin plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# Strip plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.stripplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# Swarm plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.swarmplot(data=tips, x="day", y="total_bill")
plt.show()
```

# Distribution plot

```python
# Histogtam
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=tips, x="total_bill")
plt.show()
```

```python
# Kernel Density Estimate plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.kdeplot(data=tips, x="total_bill")
plt.show()
```

```python
# Distribution plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(data=tips, x="total_bill", kind="kde")
plt.show()
```

```python
# Empirical cumulative distribution function plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.ecdfplot(data=tips, x="total_bill")
plt.show()
```

```python
# Rug plot
import seaborn as sns
import matplotlib.pyplot as plt

sns.rugplot(data=tips, x="total_bill")
plt.show()
```

# Matrix plots

```python
# Heatmap
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.heatmap(flights_pivot, annot=True, fmt="d", cmap="YlGnBu")
plt.show()
```

```python
# Cluster map
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.clustermap(flights_pivot, cmap="viridis", standard_scale=1)
plt.show()
```

# Joint plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Loading dataset
data = sns.load_dataset("attention")

# draw joinplot with hex kind
sns.jointplot(x = "solutions", y = "score", kind = "hex", data = data)

#show plot
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


# Loading dataset
data = sns.load_dataset("mpg")

# Draw jointplot with scatter kind
sns.jointplot(x = "mpg", y = "acceleration", kind = "scatter", data = data)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("exercise")

# Draw joinplot wint kde kind
sns.jointplot(x = "id", y = "pulse", kind = "kde", data = data)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("titanic")

# Draw jointplot with reg kind
sns.jointplot(x = "age", y = "fare", kind = "reg", data = data, dropna = True)

plt.show()
```
